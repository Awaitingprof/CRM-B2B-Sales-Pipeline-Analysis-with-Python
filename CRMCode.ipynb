```python
missing_by_stage = sales_pipeline.groupby("deal_stage").agg(
    total_opportunities=("opportunity_id", "count"),
    missing_account=("account", lambda x: x.isna().sum()),
    missing_engage_date=("engage_date", lambda x: x.isna().sum()),
    missing_close_date=("close_date", lambda x: x.isna().sum()),
    missing_close_value=("close_value", lambda x: x.isna().sum())
)

display(missing_by_stage)
```

```python
account_missing_by_agent = (
    sales_pipeline.groupby("sales_agent")
    .agg(
        total_opportunities=("opportunity_id", "count"),
        missing_accounts=("account", lambda x: x.isna().sum())
    )
)

account_missing_by_agent["missing_account_pct"] = (
    account_missing_by_agent["missing_accounts"]
    / account_missing_by_agent["total_opportunities"]
    * 100
)

display(
    account_missing_by_agent
    .sort_values("missing_account_pct", ascending=False)
)
```

```python
account_missing_by_product = (
    sales_pipeline.groupby("product")
    .agg(
        total_opportunities=("opportunity_id", "count"),
        missing_accounts=("account", lambda x: x.isna().sum())
    )
)

account_missing_by_product["missing_account_pct"] = (
    account_missing_by_product["missing_accounts"]
    / account_missing_by_product["total_opportunities"]
    * 100
)

display(
    account_missing_by_product
    .sort_values("missing_account_pct", ascending=False)
)
```

```python
def iqr_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    return ((series < lower) | (series > upper)).sum()

outlier_summary = pd.DataFrame({
    "Variable": [
        "Account Revenue",
        "Employees",
        "Product Sales Price",
        "Won Deal Value"
    ],
    "Outlier Count": [
        iqr_outliers(accounts_clean["revenue"]),
        iqr_outliers(accounts_clean["employees"]),
        iqr_outliers(products_clean["sales_price"]),
        iqr_outliers(
            sales_pipeline_clean.loc[
                sales_pipeline_clean["deal_stage"] == "Won",
                "close_value"
            ]
        )
    ]
})

display(outlier_summary)
```
